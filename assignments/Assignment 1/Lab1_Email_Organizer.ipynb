# Lab 1 — Agent vs Chatbot
## Email Organizer Agent
This notebook demonstrates clarification, planning, observation and revision. It does not access Gmail or send emails.

```python
from lab1_agent import EmailOrganizerAgent
agent = EmailOrganizerAgent()
```

## 1. User request and clarification
Enter a vague request. The agent asks three clarification questions.

```python
request = input('Describe your email-organizing request: ').strip()
while not request:
    request = input('Please enter a request: ').strip()
details = agent.clarify(request)
```

## 2. Generate the editable plan

```python
plan = agent.create_plan(details)
plan.display()
```

## 3. Observe feedback and revise
Examples: `add step: Group student queries separately`, `replace step 4 with Extract only confirmed deadlines`, `remove step 5`, or `change goal: Prepare a weekly faculty inbox report`.

```python
while True:
    feedback = input('Feedback (Enter to accept): ').strip()
    if not feedback:
        break
    try:
        agent.revise(plan, feedback)
        plan.display()
    except ValueError as error:
        print(error)
```

## 4. Final plan

```python
plan.display()
print('\nNotebook demonstration completed. No email was sent.')
```
